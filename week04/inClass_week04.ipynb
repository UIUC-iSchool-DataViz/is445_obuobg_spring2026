```python
# import usual things
import matplotlib.pyplot as plt
import matplotlib
import numpy as np
import pandas as pd
```

```python
buildings = pd.read_csv('https://github.com/UIUC-iSchool-DataViz/is445_data/raw/main/building_inventory.csv',
                       na_values={'Square Footage':0, 
                                 'Year Acquired':0,
                                 'Year Constructed':0,
                                 'Total Floors':0})
```

```python
#buildings.columns
```

```python
stats = buildings.groupby('Year Acquired')['Square Footage'].describe()
stats
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.plot(stats['max'], marker='.', linewidth=2.0)
ax.set_xlabel('Year')
ax.set_ylabel('Square Footage')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.plot(stats['max'], marker='.', linewidth=2.0)
ax.set_xlabel('Year')
ax.set_ylabel('Square Footage')
ax.set_yscale('log')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.plot(stats['max'], marker='.', linewidth=2.0, label='Max')
ax.plot(stats['min'], marker='.', linewidth=2.0, label='Min')
ax.set_xlabel('Year')
ax.set_ylabel('Square Footage')
ax.set_yscale('log')
ax.legend()
plt.show()
```

```python
# Step 2: make a function with this parameter as an input
#plotting_style = 'ggplot'
def make_plot(plotting_style):
    with plt.style.context(plotting_style): # use a with statement so we don't change plotting style for other plots
        # CMD/CTRL + ] 
        fig, ax = plt.subplots(figsize=(10,2))
        ax.plot(stats['max'], marker='.', linewidth=2.0, label='Max')
        ax.plot(stats['min'], marker='.', linewidth=2.0, label='Min')
        ax.set_xlabel('Year')
        ax.set_ylabel('Square Footage')
        ax.set_yscale('log')
        ax.legend()
        plt.show()
```

```python
# Step 4: decorate the function with ipywidgets.interact with the variable as an input
@ipywidgets.interact(plotting_style=plt.style.available)
def make_plot(plotting_style):
    with plt.style.context(plotting_style): # use a with statement so we don't change plotting style for other plots
        # CMD/CTRL + ] 
        fig, ax = plt.subplots(figsize=(10,2))
        ax.plot(stats['max'], marker='.', linewidth=2.0, label='Max')
        ax.plot(stats['min'], marker='.', linewidth=2.0, label='Min')
        ax.set_xlabel('Year')
        ax.set_ylabel('Square Footage')
        ax.set_yscale('log')
        ax.legend()
        plt.show()
```

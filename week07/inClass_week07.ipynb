## A few items with bqplot (before moving on)

```python
import bqplot
```

```python
# 1. data
shape_data = bqplot.topo_load('map_data/WorldMap.json') # map_data?
```

```python
# 2. scales/axis
sc_geo = bqplot.Mercator() # we know this is a bad projection!
```

```python
#. 3 Marks
#bqplot.Map?
shape_map = bqplot.Map(map_data = shape_data, scales={'projection':sc_geo})
```

```python
# no interactivity

# 5. put it together as a figure
fig = bqplot.Figure(marks=[shape_map], fig_margin={'top':0,'bottom':0,'left':0,'right':0})
fig
```

```python
# 1. data
shape_data = bqplot.topo_load('map_data/USStatesMap.json') # map_data?

# 2. scales/axis
sc_geo = bqplot.AlbersUSA() # we know this is a bad projection!

#. 3 Marks
shape_map = bqplot.Map(map_data = shape_data, scales={'projection':sc_geo})

# no interactivity

# 5. put it together as a figure
fig = bqplot.Figure(marks=[shape_map], fig_margin={'top':0,'bottom':0,'left':0,'right':0})
fig
```

```python
# 1. data
shape_data = bqplot.topo_load('map_data/WorldMap.json') # map_data?

# 2. scales/axis
sc_geo = bqplot.Gnomonic() # we know this is a bad projection!

#. 3 Marks
shape_map = bqplot.Map(map_data = shape_data, scales={'projection':sc_geo})

# no interactivity

# 5. put it together as a figure
fig = bqplot.Figure(marks=[shape_map], fig_margin={'top':0,'bottom':0,'left':0,'right':0})
fig
```

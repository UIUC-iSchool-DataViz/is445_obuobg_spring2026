```python
import traitlets
```

```python
class MyTraitletObject(traitlets.HasTraits):
    name = traitlets.Unicode('unnamed') # trailet-specific way of saying "str"
    age = traitlets.Int(0) # traitlets-specific way of saying "int"
```

```python
my_t_obj = MyTraitletObject()
```

```python
my_t_obj.name, my_t_obj.age
```

```python
def name_changed(change):
    print(change)

my_t_obj.observe(name_changed, ['name'])
```

```python
my_t_obj.name = 'Bob'
```

```python
def name_changed2(change):
    print('Hello, my name is', change['new'], 'and I am', change['owner'].age, 'years old.')

my_t_obj.unobserve(name_changed, ['name'])
my_t_obj.observe(name_changed2, ['name'])
```

```python
my_t_obj.name = 'Linda'
```

```python
# 1. Make/read in data
x = np.arange(100)
y = np.random.random(100)

# 2. Scales
x_sc = bqplot.LinearScale()
y_sc = bqplot.LinearScale()

# 3. Axis
ax_x = bqplot.Axis(scale=x_sc, label='X Value')
ax_y = bqplot.Axis(scale=y_sc, label='Y Value', orientation='vertical') 

# 4. Marks (sometimes this swaps position with Axis)
lines = bqplot.Lines(x=x,y=y,scales={'x':x_sc, 'y':y_sc})

# 5. around here interactivity if we had it
pz = bqplot.interacts.PanZoom(scales={'x':[x_sc], 'y':[y_sc]})

# 6. put it all together into a figure
fig = bqplot.Figure(marks=[lines], axes=[ax_x, ax_y], interaction=pz)
fig
```

```python
#1. read/make data
x = np.random.random(100)
y = np.random.random(100)

# 2. scales
x_sc = bqplot.LinearScale()
y_sc = bqplot.LinearScale()

# 3. Axis
ax_x = bqplot.Axis(scale=x_sc, label='X')
ax_y = bqplot.Axis(scale=y_sc, label='Y', orientation='vertical')

# 4. marks
scatters = bqplot.Scatter(x=x,y=y, scales={'x':x_sc, 'y':y_sc})

# 5. interactivity!
selector = bqplot.interacts.FastIntervalSelector(scale=x_sc, marks=[scatters])

# 6. figure
fig = bqplot.Figure(marks=[scatters], axes=[ax_x,ax_y], interaction=selector)
fig
```

```python
selector.selected
```

```python
scatters.selected_style = {'fill':'red', 'stroke':'yellow'}
```

## 2D Heatmap (random data)

```python
# 1. data
data = np.random.random((10,10))
#data.shape

# 2. scales
#bqplot.ColorScale?
col_sc = bqplot.ColorScale()

# 3. axis

# 4. mark
#bqplot.GridHeatMap?
heat_map = bqplot.GridHeatMap(color=data, scales={'color':col_sc})

# 5. interactivity

# 6. all together now
fig = bqplot.Figure(marks=[heat_map])
fig
```

```python
# 1. data
data = np.random.random((10,10))
#data.shape

# 2. scales
col_sc = bqplot.ColorScale(scheme='Greens')

# 3. axis

# 4. mark
heat_map = bqplot.GridHeatMap(color=data, scales={'color':col_sc})

# 5. interactivity

# 6. all together now
fig = bqplot.Figure(marks=[heat_map])
fig
```

```python
# 1. data
data = np.random.random((10,10))
#data.shape

# 2. scales
col_sc = bqplot.ColorScale(scheme='Greens')
x_sc = bqplot.OrdinalScale() # ordered bins
y_sc = bqplot.OrdinalScale()

# 3. axis
x_ax = bqplot.Axis(scale=x_sc, label='Column Bin Number')
y_ax = bqplot.Axis(scale=y_sc, label='Row Bin Number', orientation='vertical')

# 4. mark
heat_map = bqplot.GridHeatMap(color=data, scales={'color':col_sc, 'row':y_sc, 'column':x_sc})

# 5. interactivity

# 6. all together now
fig = bqplot.Figure(marks=[heat_map], axes=[x_ax, y_ax])
fig
```

```python
import traitlets
```

```python
class MyTraitletObject(traitlets.HasTraits):
    name = traitlets.Unicode('unnamed') # trailet-specific way of saying "str"
    age = traitlets.Int(0) # traitlets-specific way of saying "int"
```

```python
my_t_obj = MyTraitletObject()
```

```python
my_t_obj.name, my_t_obj.age
```

```python
def name_changed(change):
    print(change)

my_t_obj.observe(name_changed, ['name'])
```

```python
my_t_obj.name = 'Bob'
```

```python
def name_changed2(change):
    print('Hello, my name is', change['new'], 'and I am', change['owner'].age, 'years old.')

my_t_obj.unobserve(name_changed, ['name'])
my_t_obj.observe(name_changed2, ['name'])
```

```python
my_t_obj.name = 'Linda'
```

## 3D random heatmap

```python
data3d = np.random.random((10,10,20))
data3d.shape
```

```python
np.sum(data3d, axis=2).shape
```

```python
# 1. data
# data not updating

# 2. scales
col_sc = bqplot.ColorScale(scheme='Greens')
x_sc = bqplot.OrdinalScale() # ordered bins
y_sc = bqplot.OrdinalScale()

# 3. axis
x_ax = bqplot.Axis(scale=x_sc, label='Column Bin Number')
y_ax = bqplot.Axis(scale=y_sc, label='Row Bin Number', orientation='vertical')
col_ax = bqplot.ColorAxis(scale=col_sc, orientation='vertical', side='right')

# 4. mark
heat_map = bqplot.GridHeatMap(color=np.sum(data3d, axis=2), 
                              scales={'color':col_sc, 'row':y_sc, 'column':x_sc},
                             interactions={'click':'select'},
                             anchor_style={'fill':'magenta'})

# 5. interactivity
def on_selected(change):
    if len(change['owner'].selected) == 1: # only allow a single selection
        i,j = change['owner'].selected[0] # row number, column number (hint, if something isn't working, flip these!)
        myLabel.value = 'Data value (sum) = ' + str(np.sum(data3d[i,j,:])) # if something looks wrong, is i/j flipped?

heat_map.observe(on_selected, 'selected') # (name_changed, ['name'])

# 6. all together now
fig = bqplot.Figure(marks=[heat_map], axes=[x_ax, y_ax, col_ax])

# 7. dashboard it!
dashboard = ipywidgets.VBox([myLabel, fig])
dashboard
```

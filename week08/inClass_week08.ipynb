#### First, quick mock-up of what we want things to look like using matplotlib

````python
import pandas as pd
````

````python
df = pd.read_csv("https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/mobility.csv")
````

````python
df.head()
````

````python
df.columns
````

Think ahead -- we want a dashboard with:
1. on the "driver"/left side -- a heatmap of State (y-axis) and Student-to-teacher ratio (x-axis)
2. on the responder/right side -- a plot of the Mobility score for a selection of State/Student-to-teacher ratio

````python
df['Mobility'].plot(kind='hist')
````

````python
df['State'].unique()
````

````python
df['Student_teacher_ratio'].plot(kind='hist', bins=10)
````

````python
import numpy as np
````

````python
bins = np.linspace(df['Student_teacher_ratio'].min(), df['Student_teacher_ratio'].max(), 10)
````

````python
bins
````

````python
table = df.pivot_table(index='State', columns=pd.cut(df['Student_teacher_ratio'],bins), aggfunc='size')
````

````python
table
````

````python
import matplotlib.pyplot as plt
````

````python
fig, ax = plt.subplots(figsize=(10,8))
# extent (xmin, ymin, xmax, ymax)
extent = [bins.min(), bins.max(), 0, len(table.index)]
ax.imshow(table.values, cmap='hot', extent=extent)
ax.set_yticks(range(len(table.index))) # one tick mark for each 50 states
ax.set_yticklabels(table.index) # each state name
plt.show()
````

### Making a dashboard with Altair and vega-lite syntax

```javascript
{
  "$schema": "https://vega.github.io/schema/vega-lite/v6.json",
  "data": {"url": "data/movies.json"},
  "mark": "bar",
  "encoding": {
    "x": {
      "bin": true,
      "field": "IMDB Rating"
    },
    "y": {"aggregate": "count"}
  }
}
```

````python
chart2 = alt.Chart.from_dict(
{
  #"$schema": "https://vega.github.io/schema/vega-lite/v6.json",
    # 1. Data
    "data": {"url": "https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/mobility.csv"},
    # 2. Mark
    "mark": "bar",
    # 3. Encodings (scales + axis)
    "encoding": {
        "x": {
            #"bin": true,
            "bin": True,
            "field": "Mobility", 
            "type":"quantitative", 
            "axis":{"title":"Mobility Score"}
        },
        "y": {"aggregate": "count", "type":"quantitative", "axis":{"title":"Frequency"}}
  }
})
chart2
````

````python
chart1 = alt.Chart.from_dict(
{
    # 1. Data
    "data": {"url": "https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/mobility.csv"},
    # 2. Mark
    "mark": "rect",
    # 3. Encodings (scales + axis)
    "encoding": {
        "x": {"field":"Student_teacher_ratio", "type":"quantitative"},
        "y": {"field":"State", "type":"ordinal"}
  }
})
chart1
````

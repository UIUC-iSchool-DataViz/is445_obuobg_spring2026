```python
im = Image.open('stitch_reworked.png')
```

```python
type(im)
```

```python
import numpy as np
```

```python
im_data = np.array(im)
```

```python
im_data.shape # 483 rows, 430 columns, 4 color channels -- R=red, G=green, B=blue, A=alpha/opacity
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.hist(im_data[:,:,0]) # [:,:,0] == all pixels in rows, columns in the Red color channel
plt.show()
```

```python
im_data[:,:,0].shape
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.hist(im_data[:,:,0].flatten()) # [:,:,0] == all pixels in rows, columns in the Red color channel
# instead of passing a 2D thing, we are passing a 1d-thing
plt.show()
```

```python
im_data[:,:,0].flatten().shape
```

```python
np.unique(im_data[:,:,0])
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.hist(im_data[:,:,0].flatten(), bins=256) # [:,:,0] == all pixels in rows, columns in the Red color channel
# instead of passing a 2D thing, we are passing a 1d-thing
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.hist(im_data[:,:,0].flatten(), bins=256, color='red') # [:,:,0] == all pixels in rows, columns in the Red color channel
ax.set_xlabel('Red channel bins')
ax.set_ylabel('# of pixels per bin')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.hist(im_data[:,:,0].flatten(), bins=256, color='red') # [:,:,0] == Red channel
ax.hist(im_data[:,:,1].flatten(), bins=256, color='green') # [:,:,1] == Green channel
ax.set_xlabel('Channel bins')
ax.set_ylabel('# of pixels per bin')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.hist(im_data[:,:,0].flatten(), bins=256, color='red') # [:,:,0] == Red channel
ax.hist(im_data[:,:,1].flatten(), bins=256, color='green') # [:,:,1] == Green channel
ax.hist(im_data[:,:,2].flatten(), bins=256, color='blue') # [:,:,2] == Blue channel
ax.hist(im_data[:,:,3].flatten(), bins=256, color='gray') # [:,:,3] == Alpha/opacity channel
ax.set_xlabel('Channel bins')
ax.set_ylabel('# of pixels per bin')
plt.show()
```
